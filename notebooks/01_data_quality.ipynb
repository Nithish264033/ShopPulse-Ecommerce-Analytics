```python

import pandas as pd
import numpy as np
import os

print("ShopPulse Data Quality Analysis")
print("Libraries imported successfully!")
```

```python

# Define the dataset folder
data_path = "../data/raw/"

# Load the main datasets
customers = pd.read_csv(data_path + "olist_customers_dataset.csv")
orders = pd.read_csv(data_path + "olist_orders_dataset.csv")
order_items = pd.read_csv(data_path + "olist_order_items_dataset.csv")
products = pd.read_csv(data_path + "olist_products_dataset.csv")
sellers = pd.read_csv(data_path + "olist_sellers_dataset.csv")
payments = pd.read_csv(data_path + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(data_path + "olist_order_reviews_dataset.csv")

print("All datasets loaded successfully!")
```

```python

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Sellers": sellers,
    "Payments": payments,
    "Reviews": reviews
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python

# Check missing values in each dataset

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values found.")
    else:
        print(missing)
```

```python

# Calculate missing value percentage

for name, df in datasets.items():
    print(f"\n--- {name} ---")

    missing_percentage = (
        df.isnull().sum() / len(df) * 100
    ).round(2)

    print(missing_percentage[missing_percentage > 0])
```

```python

# Check duplicate records in each dataset

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()

    print(f"{name}: {duplicate_count} duplicate rows")
```

```python

# Check duplicate IDs in important datasets

id_columns = {
    "Customers": ("customers", "customer_id"),
    "Orders": ("orders", "order_id"),
    "Products": ("products", "product_id"),
    "Sellers": ("sellers", "seller_id")
}

for name, (variable, column) in id_columns.items():
    df = datasets[name]
    print(
        f"{name}: {df[column].duplicated().sum()} duplicate {column} values"
    )
```

```python

# Display column names and data types

for name, df in datasets.items():
    print(f"\n========== {name} ==========")
    print(df.dtypes)
```

```python

# Generate a summary of all datasets

for name, df in datasets.items():
    print(f"\n========== {name} ==========")
    display(df.info())
```

```python

# Check order date columns

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(orders[column], errors="coerce")

print(orders[date_columns].dtypes)
```

```python

# Create a data quality summary

quality_summary = []

for name, df in datasets.items():
    quality_summary.append({
        "Dataset": name,
        "Total Rows": df.shape[0],
        "Total Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_df = pd.DataFrame(quality_summary)

display(quality_df)
```
